```python
%sql
CREATE TABLE IF NOT EXISTS databricksbootcamp.aerosurf.asset_status (
    asset_sk BIGINT GENERATED ALWAYS AS IDENTITY,
    asset_id STRING,
    beach STRING,
    category STRING,
    condition STRING,
    hire_status STRING,
    inspected_ts TIMESTAMP,
    defect_notes STRING,
    start_ts TIMESTAMP,
    end_ts TIMESTAMP,
    is_current BOOLEAN
) TBLPROPERTIES(delta.enableChangeDataFeed = true)
```

```python
%sql
MERGE INTO databricksbootcamp.aerosurf.asset_status t
USING (
SELECT NULL as MergedKey, * FROM databricksbootcamp.aerosurf.raw_inspections
UNION ALL 
SELECT i.asset_id as MergedKey, i.* FROM databricksbootcamp.aerosurf.raw_inspections i
INNER JOIN databricksbootcamp.aerosurf.asset_status e ON i.asset_id = e.asset_id
) s
ON t.asset_id = s.MergedKey AND t.is_current = true 

WHEN MATCHED THEN UPDATE SET is_current = false, end_ts = current_timestamp()
WHEN NOT MATCHED THEN INSERT (asset_id, beach, category, condition, hire_status, inspected_ts, defect_notes, start_ts, end_ts, is_current)
VALUES (s.asset_id, s.beach, s.category, s.condition, s.hire_status, s.inspected_ts, s.defect_notes, current_date(), NULL, true)
```
